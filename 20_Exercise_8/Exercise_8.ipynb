## 🏋️ Exercise: A simple currency helper

1. Pick any two currencies (for example, EUR to INR).
2. Call the API and read the rate from the JSON.
3. Take an amount, say 100, and calculate how much it is in the other currency.
4. Print the result clearly, for example: `100 EUR = 9050 INR`.

**Bonus:** pass the converted amount to the LLM and ask it to write a one-line travel tip that mentions the amount. Now you have combined a live API with the LLM — a small but real app. 🚀

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
```

```python
import requests

# 1. Fetch the exchange rate
try:
    response = requests.get(
        "https://api.frankfurter.app/latest",
        params={"from": "EUR", "to": "INR"},
        timeout=10
    )
    response.raise_for_status()
    data = response.json()

    rate = data["rates"]["INR"]
    amount = 100
    converted_amount = amount * rate

    print(f"{amount} EUR = {converted_amount:.2f} INR")
    print(f"Exchange rate: 1 EUR = {rate:.2f} INR")

    # 2. Pass the converted amount to Gemini
    prompt = f"""
    A traveler has {converted_amount:.2f} INR,
    converted from {amount} EUR.

    Write one short, practical travel tip that
    mentions the amount available in INR.
    """

    try:
        response = client.models.generate_content(
            model="gemini-3.8-flash",
            contents=prompt
        )

        print("\nTravel Tip:")
        print(response.text)

    except Exception as e:
        print(f"Gemini API error: {e}")

except requests.RequestException as e:
    print(f"Currency API error: {e}")

except (KeyError, ValueError) as e:
    print(f"Unexpected API response: {e}")
```
