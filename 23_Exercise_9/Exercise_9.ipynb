## Exercise

A small shop wants to store its products in a database.

1. Create a table called `products` with columns: `id`, `name` (TEXT), and `price` (INTEGER).
2. Add at least 3 products (for example: "Tea", 10).
3. Read and print all products.
4. **Bonus:** print only the products with a price greater than 20 (hint: `WHERE price > ?`).

Tip: you can reuse `shop.db` from above — a database can hold many tables.

```python
import sqlite3

conn = sqlite3.connect("shop.db")
cursor = conn.cursor()

print("Database is ready!")
```

```python
cursor.execute("""
   CREATE TABLE IF NOT EXISTS products(
        id INTEGER PRIMARY KEY,
        name TEXT,
        price INTEGER
   ) 
""")

print("Table created!")
```

```python
cursor.execute("SELECT COUNT(*) FROM products")
count = cursor.fetchone()[0]

if count == 0:
    more_products = [
        ("Tea", 10),
        ("Pasta", 22),
        ("Sandwich", 20),
        ("Burger", 25),
    ]

    cursor.executemany(
        "INSERT INTO products (name, price) VALUES (?, ?)",
        more_products
    )

    conn.commit()
    print("Added", len(more_products), "products.")
else:
    print("Products already exist. Skipping insertion.")
```

```python
cursor.execute("SELECT * FROM products")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("SELECT name, price FROM products WHERE price > ?", (20,))
twenty = cursor.fetchall()

print("Products more than 20:")
for name, price in twenty:
    print("-", name, "(price:", price, ")")
```
