````python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
````

````python
from google.genai import types
````

````python
#Start a fresh conversation
conversation = []

# Give the chatbot a personality
config = types.GenerateContentConfig(
    system_instruction = (
        "You are a witty movie buff who loves making puns. "
        "Remember the user's movie preferences from the conversation. "
        "Recommend movies that match their tastes and avoid genres "
        "they dislike. Explain your recommendations clearly."
    )
)
````

````python
def chat(user_message):
    # Add the user's message to memory
    conversation.append({
        "role": "user",
        "parts": [{"text": user_message}]
    })

    try:
        # Send the entire conversation to Gemini
        response = client.models.generate_content(
            model="gemini-3.8-flash",
            contents=conversation,
            config=config
        )

        reply = response.text

        # Save the chatbot's reply to memory
        conversation.append({
            "role": "model",
            "parts": [{"text": reply}]
        })

        print("Bot:", reply)
        return reply

    except Exception as e:
        print("Something went wrong:", e)
        return None
````

````python
# 4. Bonus 1: Reset the conversation
def reset():
    conversation.clear()
    print("New chat started! Memory cleared.")
````

````python
# 5. Test the chatbot
chat("I love sci-fi and thrillers, but I can't stand horror movies.")
````

````python
chat("Suggest me a movie to watch tonight.")
````

````python
chat("Why did you pick that one?")
````
