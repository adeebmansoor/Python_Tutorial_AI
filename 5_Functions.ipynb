# Functions in Python

```python
def calculate_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin
```

```python
calculate_margin(50, 10)
```

### Another way

```python
revenue = 60
expenses = 15

calculate_margin(revenue, expenses)
```

### Multiple return values

```python
def calculate_profit_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return profit, margin

revenue = 50
expenses = 10

profit, margin = calculate_profit_margin(revenue, expenses)
```

```python
profit
```

```python
margin
```

```python
response = calculate_profit_margin(revenue, expenses)
type(response)
```

### Returning as a list

```python
def calculate_profit_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return [profit, margin]
```

```python
response = calculate_profit_margin(revenue, expenses)
type(response)
```

## Type hints (labeling your inputs and outputs)

Look at our function: `def calculate_margin(revenue, expenses)`. Nothing tells you — or the next developer — what `revenue` is meant to be. A number? A list? Text?

Python lets you add optional **type hints**: tiny labels saying what type you expect.
It doesn't mean we are creating funtions with datatypes and return types, it is just for giving hints.

```python
def calculate_profit_margin(revenue: float, expenses: float) -> float:
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return profit, margin
```

Two new bits:
- `revenue: float` — *"this input should be a float"*
- `-> float` — *"this function returns a float"*

> ⚠️ **The catch:** type hints are just labels for humans (and tools). Python does **not** enforce them. Watch — we promise an `int`, hand it a string, and Python happily runs anyway:

```python
def double(x: int) -> int:
    return x * 2

print(double("ha"))   # we said int... gave a string... Python shrugs -> 'haha'
```

So why bother to type hints? Because hints make your code **self-documenting**, your editor autocompletes better, and AI coding assistants write *far* better code when they can see the types.
