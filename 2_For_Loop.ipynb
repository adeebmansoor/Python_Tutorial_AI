# For Loop in Python

```python
revenues = [50, 60, 55]
expenses = [10, 30, 35]
```

```python
for rev in revenues:
    print(rev)
```

```python
for rev in revenues:
    print("Revenues:", rev)
```

```python
for rev in revenues:
    print(f"Revenues: {rev}")
```

### Profit margins for different revenues and expenses

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(margin)
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(f"Margin Q{i+1} : {margin:.2f}")
```

### Storing answers in another list

```python
margins = []
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

margins
```
