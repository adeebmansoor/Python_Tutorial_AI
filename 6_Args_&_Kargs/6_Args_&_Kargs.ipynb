# Args and Kargs

```python
# This is a normal funtion
def calculate_expenses(exp1, exp2):
    return exp1 + exp2

calculate_expenses(10, 20)
```

### We define args using single asterisk(*)
we can add multiple arguments without actually defining them in the funtion

```python
def calculate_expenses(*args):
    print(args)
    return sum(args)

calculate_expenses(10, 20, 30)
```

### What if we want to define a function with arguments carrying statings?

```python
def total_expenses(rent, bill):
    return rent + bill

total_expenses(rent=5000, bill=1000)
```

## For these type of functions we use Kargs arguments with double asterisk(**)
### Values in kargs are stored as dictionary

```python
def total_expenses(**kargs):
    total = 0
    for key, data in kargs.items():
        total += data
    return total

total_expenses(rent=4000, bill=1000, expenses=2000)
```
