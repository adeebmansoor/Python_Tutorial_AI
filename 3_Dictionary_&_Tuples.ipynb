# Dictionary in Python

```python
fin = {
    "revenue": 50,
    "expenses": 10
}
fin
```

```python
type(fin)
```

```python
profit = fin["revenue"] - fin["expenses"]
profit
```

```python
fin['profit'] = fin["revenue"] - fin["expenses"]
fin
```

## List of Dictionaries

```python
financials = [
    {"revenue": 50, "expenses": 10},  #q1
    {"revenue": 60, "expenses": 30},  #q2
    {"revenue": 55, "expenses": 35}   #q3
]
financials
```

```python
for fin in financials:
    profit = fin["revenue"] - fin["expenses"]
    fin["margin"] = profit * 100 / fin["revenue"]

financials
```

## Nested Dictionaries

```python
financials = {
    "Q1" : {"revenue": 50, "expenses": 10},  #q1
    "Q2" : {"revenue": 60, "expenses": 30},  #q2
    "Q3" : {"revenue": 55, "expenses": 35}   #q3
}

financials
```

```python
for quarter,data in financials.items():
    print("Quarters:", quarter)
    print("Data:", data)
```

```python
for quarter,data in financials.items():
    margin = (data["revenue"] - data["expenses"]) * 100 / data["revenue"]
    print(f"{quarter}: {margin:.2f}%")
```

```python
financials = {
    "Q1" : {"revenue": 50, "expenses": 10},  #q1
    "Q2" : {"revenue": 60, "expenses": 30},  #q2
    "Q3" : {"revenue": 55, "expenses": 35},  #q3
    "Q4" : {"revenue": -10, "expenses": 35}  #q4
}

financials
```

## If Condition

```python
for quarter,data in financials.items():
    if data["revenue"] < 0:
        print(f"Invalid revenue for {quarter}, Skipping.")
        continue
    margin = (data["revenue"] - data["expenses"]) * 100 / data["revenue"]
    print(f"{quarter}: {margin:.2f}%")
```

# Tuples in python

### Tuples and list behaves the same but one difference between them is that tuples are immutable where as lists are mutable in nature 

***A tuple is like your mom's recipe — fixed, sacred, don't you dare change the masala. A list is like your Swiggy order — add, remove, modify till the delivery guy arrives.***

```python
purple = (128, 0, 128)
purple[0]
```

```python
purple[1]
```

### Tuples are immutable hence reinitializing them with new values is not possible

```python
# purple[1] = 128
```

```python
r, g, b = purple
print(r)
print(g)
print(b)
```

### Mixing color grades using tuples in python

```python
yellow = (255, 255, 0)
mixed = (purple[0] + yellow[0])/2, (purple[1] + yellow[1])/2, (purple[2] + yellow[2])/2

mixed
```

```python
mixed = (purple[0] + yellow[0])//2, (purple[1] + yellow[1])//2, (purple[2] + yellow[2])//2

mixed
```
